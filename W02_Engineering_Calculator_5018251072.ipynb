```python
# Week 2 Assignment task 1
## Engineering Unit and Electrical Calculator
# Nama : stevanus mahendra lisvantono
# NRP : 5018251072
# Kelas : pemprogaman komputer a

# meminta input dari pengguna untuk kecepatan dalam km/h
speed_kmh = float(input("Enter speed [km/h]: "))

# mengonversi kecepatan dari km/h menjadi m/s
speed_ms = speed_kmh / 3.6

# menampilkan hassil konversi dengan format dua angka desimal
print(f"Speed = {speed_ms:.2f} m/s")
```

```python
# Week 2 Assignment task 2
## Engineering Unit and Electrical Calculator
# Nama : stevanus mahendra lisvantono
# NRP : 5018251072
# Kelas : pemprogaman komputer a

# Meminta input daya dalam kW dari pengguna
power_kw = float(input("Power [kW]: "))

# Mengonversi daya dari kW menjadi W menggunakan persamaan P_W = 1000 * P_kW
power_w = power_kw * 1000

# Menampilkan hasil dengan format 2 angka desimal sesuai contoh output
print(f"Power = {power_w:.2f} W")
```

```python
# Week 2 Assignment task 3
## Engineering Unit and Electrical Calculator
# Nama : stevanus mahendra lisvantono
# NRP : 5018251072
# Kelas : pemprogaman komputer a

# Meminta input dari pengguna
voltage = float(input("Masukkan tegangan (V): "))
current = float(input("Masukkan arus (A): "))
time = float(input("Masukkan waktu operasi (jam): "))

# Menghitung daya (P = V * I)
power = voltage * current

# Menghitung energi (E = P * t)
energy = power * time

# Menampilkan output menggunakan f-string dengan format 2 angka desimal
print("\n===== ELECTRICAL CALCULATION =====")
print(f"Voltage = {voltage:.2f} V")
print(f"Current = {current:.2f} A")
print(f"Power = {power:.2f} W")
print(f"Operating Time = {time:.2f} hour")
print(f"Energy = {energy:.2f} Wh")
```

```python
# Week 2 Assignment task 4
## Engineering Unit and Electrical Calculator
# Nama : stevanus mahendra lisvantono
# NRP : 5018251072
# Kelas : pemprogaman komputer a

# Batas nilai yang ditentukan
power_limit = 200
voltage_limit = 24

# Membuat variabel boolean dengan membandingkan hasil Task 3 dengan batas
is_high_power = power > power_limit
is_high_voltage = voltage > voltage_limit

# Menampilkan hasil evaluasi kondisi logika
print(f"High power : {is_high_power}")
print(f"High voltage : {is_high_voltage}")
print(f"Both conditions : {is_high_power and is_high_voltage}")
print(f"Any condition : {is_high_power or is_high_voltage}")
```

1. Apa perbedaan int dan float?

int (integer) digunakan untuk menyimpan bilangan bulat tanpa koma (contoh: 1, -5, 100).float (floating point) digunakan untuk menyimpan bilangan pecahan atau desimal (contoh: 3.14, 5.0, -0.5).

2. Mengapa hasil input() perlu dikonversi menggunakan float() untuk perhitungan numerik?

Fungsi input() di Python secara otomatis menyimpan apa pun yang diketik oleh pengguna sebagai tipe data teks (string). Jika teks tersebut berisi angka dan kita ingin menjumlahkan atau mengalikannya, teks tersebut harus diubah terlebih dahulu menjadi tipe data angka (float atau int). Jika tidak dikonversi, program akan menghasilkan error atau menggabungkan teksnya saja.

3. Apa perbedaan operator = dan == dalam Python?

= adalah assignment operator (operator penugasan). Fungsinya untuk memasukkan atau menetapkan sebuah nilai ke dalam suatu variabel (contoh: x = 5).

== adalah comparison operator (operator perbandingan). Fungsinya untuk memeriksa apakah nilai di sebelah kiri sama persis dengan nilai di sebelah kanan, yang nantinya akan menghasilkan nilai Boolean True atau False (contoh: x == 5).

4. Apa fungsi operator and dan or?

and: Menghasilkan nilai True HANYA JIKA seluruh kondisi yang dievaluasi bernilai True. Jika ada satu saja yang False, hasilnya False.

or: Menghasilkan nilai True JIKA SALAH SATU (atau semua) kondisi yang dievaluasi bernilai True. Hasilnya hanya False jika seluruh kondisinya False.
