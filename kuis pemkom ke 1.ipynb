```python
score = 85
if score >= 90:
  grade = "A"
elif score >= 80:
  grade = "B"
elif score >= 70:
  grade = "C"
else:
  grade = "D"

print(f"Grade Anda: {grade}")
```

```python
#nomer 4

x = float(input("Masukkan nilai pengukuran zat kimia: "))

if 0 <= x < 40:
    status = "LOW"
elif 40 <= x < 70:
    status = "NORMAL"
elif 70 <= x < 90:
    status = "HIGH"
elif 90 <= x <= 100:
    status = "CRITICAL"
else:
    status = "INVALID"

print(f"Nilai pengukuran: {x}")
print(f"Status bahaya: {status}")
```

```python
#nomer 3

temperature = 85.0
pump_running = False

if temperature > 80.0 and pump_running == False:
    print("ALARM")
else:
    print("NORMAL")
```
